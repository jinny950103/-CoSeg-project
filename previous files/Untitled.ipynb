```python
import os
import numpy as np
import nibabel as nib
import cv2

# 路徑
input_image_dir = "Task02_Heart/imagesTr"
input_label_dir = "Task02_Heart/labelsTr"

output_image_dir = "dataset/images"
output_mask_dir = "dataset/masks"

os.makedirs(output_image_dir, exist_ok=True)
os.makedirs(output_mask_dir, exist_ok=True)

# 讀所有檔案
for file in os.listdir(input_image_dir):
    if file.endswith(".nii.gz"):
        img_path = os.path.join(input_image_dir, file)
        label_path = os.path.join(input_label_dir, file)

        img = nib.load(img_path).get_fdata()
        label = nib.load(label_path).get_fdata()

        # 取中間 slice（先簡單做）
        z = img.shape[2] // 2
        img_slice = img[:, :, z]
        label_slice = label[:, :, z]

        # normalize
        img_slice = (img_slice - img_slice.min()) / (img_slice.max() - img_slice.min()) * 255
        img_slice = img_slice.astype(np.uint8)

        name = file.replace(".nii.gz", "")

        # 存 PNG
        cv2.imwrite(f"{output_image_dir}/{name}.png", img_slice)

        # 存 NPY
        np.save(f"{output_mask_dir}/{name}.npy", label_slice)

print("Done!")
```

```python
import os
import numpy as np
import nibabel as nib
import cv2

input_image_dir = "Task02_Heart/imagesTr"
input_label_dir = "Task02_Heart/labelsTr"

output_image_dir = "output_images"
output_mask_dir = "output_masks"

os.makedirs(output_image_dir, exist_ok=True)
os.makedirs(output_mask_dir, exist_ok=True)

for file in os.listdir(input_image_dir):

    # ❗跳過垃圾檔（Mac 會有）
    if file.startswith("._"):
        continue

    # ❗只處理 nii.gz
    if not file.endswith(".nii.gz"):
        continue

    try:
        img_path = os.path.join(input_image_dir, file)
        label_path = os.path.join(input_label_dir, file)

        img = nib.load(img_path).get_fdata()
        label = nib.load(label_path).get_fdata()

    except:
        print("❌ skip broken file:", file)
        continue

    # 👉 取中間 slice（簡單 demo 用）
    z = img.shape[2] // 2
    img_slice = img[:, :, z]
    label_slice = label[:, :, z]

    # 👉 normalize 到 0~255
    img_slice = (img_slice - img_slice.min()) / (img_slice.max() - img_slice.min()) * 255
    img_slice = img_slice.astype(np.uint8)

    name = file.replace(".nii.gz", "")

    cv2.imwrite(f"{output_image_dir}/{name}.png", img_slice)
    np.save(f"{output_mask_dir}/{name}.npy", label_slice)

print("✅ Done!")
```
