```python
import os
import numpy as np
import cv2
import torch
from torch.utils.data import Dataset

class SimpleDataset(Dataset):
    def __init__(self, root):
        self.img_dir = os.path.join(root, "label")
        self.mask_dir = os.path.join(root, "mask_npy")  # ✅ 這裡

        self.img_files = sorted([f for f in os.listdir(self.img_dir) if f.endswith(".dcm")])
        self.mask_files = sorted([f for f in os.listdir(self.mask_dir) if f.endswith(".npy")])

        print("images:", len(self.img_files))
        print("masks :", len(self.mask_files))

        assert len(self.img_files) == len(self.mask_files), "❌ image/mask 數量不一致"

    def __len__(self):
        return len(self.img_files)

    def __getitem__(self, idx):
        import pydicom

        img_path = os.path.join(self.img_dir, self.img_files[idx])
        mask_path = os.path.join(self.mask_dir, self.mask_files[idx])

        # 讀 DICOM
        dcm = pydicom.dcmread(img_path)
        img = dcm.pixel_array.astype("float32")

        # 讀 npy mask
        mask = np.load(mask_path).astype("float32")

        # resize
        img = cv2.resize(img, (256, 256))
        mask = cv2.resize(mask, (256, 256))

        # binarize（很重要）
        mask = (mask == 1).astype("float32")

        # to tensor
        img = torch.tensor(img).unsqueeze(0)
        mask = torch.tensor(mask).unsqueeze(0)

        return img, mask
```

```python
import torch
import torch.nn as nn

class UNet(nn.Module):
    def __init__(self):
        super().__init__()

        def CBR(in_c, out_c):
            return nn.Sequential(
                nn.Conv2d(in_c, out_c, 3, padding=1),
                nn.ReLU(),
                nn.Conv2d(out_c, out_c, 3, padding=1),
                nn.ReLU()
            )

        self.enc1 = CBR(1, 64)
        self.pool1 = nn.MaxPool2d(2)

        self.enc2 = CBR(64, 128)
        self.pool2 = nn.MaxPool2d(2)

        self.enc3 = CBR(128, 256)
        self.pool3 = nn.MaxPool2d(2)

        self.middle = CBR(256, 512)

        self.up3 = nn.ConvTranspose2d(512, 256, 2, stride=2)
        self.dec3 = CBR(512, 256)

        self.up2 = nn.ConvTranspose2d(256, 128, 2, stride=2)
        self.dec2 = CBR(256, 128)

        self.up1 = nn.ConvTranspose2d(128, 64, 2, stride=2)
        self.dec1 = CBR(128, 64)

        self.out = nn.Conv2d(64, 1, 1)

    def forward(self, x):
        e1 = self.enc1(x)
        e2 = self.enc2(self.pool1(e1))
        e3 = self.enc3(self.pool2(e2))

        m = self.middle(self.pool3(e3))

        d3 = self.up3(m)
        d3 = torch.cat([d3, e3], dim=1)
        d3 = self.dec3(d3)

        d2 = self.up2(d3)
        d2 = torch.cat([d2, e2], dim=1)
        d2 = self.dec2(d2)

        d1 = self.up1(d2)
        d1 = torch.cat([d1, e1], dim=1)
        d1 = self.dec1(d1)

        return self.out(d1)
```

```python
from torch.utils.data import DataLoader

dataset = SimpleDataset("/workspace/-CoSeg-project/dataset")

print("dataset size:", len(dataset))

loader = DataLoader(dataset, batch_size=4, shuffle=True)

device = "cuda" if torch.cuda.is_available() else "cpu"

model = UNet().to(device)
loss_fn = torch.nn.BCEWithLogitsLoss()
opt = torch.optim.Adam(model.parameters(), lr=1e-3)

for epoch in range(20):
    total = 0

    for x, y in loader:
        x, y = x.to(device), y.to(device)

        pred = model(x)
        loss = loss_fn(pred, y)

        opt.zero_grad()
        loss.backward()
        opt.step()

        total += loss.item()

    print(f"epoch {epoch} loss {total:.4f}")
```

```python
import matplotlib.pyplot as plt

model.eval()

img, mask = dataset[40]  # 你說40附近有牙齒

img_input = img.unsqueeze(0).to(device)

with torch.no_grad():
    pred = model(img_input)

pred = torch.sigmoid(pred)[0,0].cpu().numpy()

plt.figure(figsize=(12,3))

plt.subplot(1,3,1)
plt.title("Input")
plt.imshow(img[0], cmap="gray")

plt.subplot(1,3,2)
plt.title("GT")
plt.imshow(mask[0], cmap="gray")

plt.subplot(1,3,3)
plt.title("Prediction")
plt.imshow(pred > 0.1, cmap="gray")

plt.show()
```
