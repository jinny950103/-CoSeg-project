```python
import nibabel as nib
import numpy as np
import os

nii_path = "/workspace/-CoSeg-project/dataset/clean_mask.nii"

img = nib.load(nii_path)
data = img.get_fdata()

print("shape:", data.shape)
print("max:", np.max(data))
print("min:", np.min(data))

save_dir = "/workspace/-CoSeg-project/dataset/clean_mask"
os.makedirs(save_dir, exist_ok=True)

count = 0

for i in range(data.shape[2]):

    mask = data[:, :, i]

    if np.max(mask) > 0:

        mask = (mask > 0).astype(np.float32)

        np.save(
            os.path.join(save_dir, f"{i:04d}.npy"),
            mask
        )

        count += 1

print("saved:", count)
```

```python
import os, cv2, pydicom, numpy as np

ROOT = "/workspace/-CoSeg-project/dataset"

IMG_DIR = os.path.join(ROOT, "label")
SAVE_DIR = os.path.join(ROOT, "pseudo_tooth_mask")

os.makedirs(SAVE_DIR, exist_ok=True)

files = sorted([f for f in os.listdir(IMG_DIR) if f.endswith(".dcm")])

print("dcm files:", len(files))

for i, fname in enumerate(files):

    path = os.path.join(IMG_DIR, fname)

    dcm = pydicom.dcmread(path)

    img = dcm.pixel_array.astype(np.float32)

    img_norm = (img - img.min()) / (img.max() - img.min() + 1e-8)

    mask = (img_norm > 0.7).astype(np.float32)

    kernel = np.ones((3,3), np.uint8)

    mask = cv2.morphologyEx(mask, cv2.MORPH_OPEN, kernel)
    mask = cv2.morphologyEx(mask, cv2.MORPH_CLOSE, kernel)

    np.save(
        os.path.join(SAVE_DIR, f"{i:04d}.npy"),
        mask
    )

print("DONE")
```

```python
!pip install pydicom opencv-python matplotlib -q

import os, cv2, pydicom, torch
import numpy as np
import torch.nn as nn
import matplotlib.pyplot as plt
from torch.utils.data import Dataset, DataLoader

ROOT = "/workspace/-CoSeg-project/dataset"
MASK_DIR = os.path.join(ROOT, "pseudo_tooth_mask")

# 自動找 DICOM 資料夾
possible_img_dirs = ["images", "no_label", "label"]
IMG_DIR = None

for d in possible_img_dirs:
    p = os.path.join(ROOT, d)
    if os.path.exists(p):
        files = [f for f in os.listdir(p) if f.lower().endswith(".dcm")]
        if len(files) > 0:
            IMG_DIR = p
            break

print("IMG_DIR:", IMG_DIR)
print("MASK_DIR:", MASK_DIR)

assert IMG_DIR is not None, "找不到 DICOM 影像資料夾"
assert os.path.exists(MASK_DIR), "找不到 clean_mask 資料夾"


class CleanMaskDataset(Dataset):
    def __init__(self, img_dir, mask_dir):
        self.img_dir = img_dir
        self.mask_dir = mask_dir

        self.img_files = sorted([f for f in os.listdir(img_dir) if f.lower().endswith(".dcm")])
        self.mask_files = sorted([f for f in os.listdir(mask_dir) if f.endswith(".npy")])

        print("images:", len(self.img_files))
        print("clean masks:", len(self.mask_files))

        assert len(self.img_files) > 0
        assert len(self.mask_files) > 0

    def __len__(self):
        return len(self.mask_files)

    def __getitem__(self, idx):
        mask_name = self.mask_files[idx]

        # clean_mask 是 0040.npy 這種，所以用檔名當 image index
        slice_idx = int(mask_name.replace(".npy", ""))

        img_name = self.img_files[slice_idx]
        img_path = os.path.join(self.img_dir, img_name)
        mask_path = os.path.join(self.mask_dir, mask_name)

        dcm = pydicom.dcmread(img_path)
        img = dcm.pixel_array.astype(np.float32)

        img = (img - img.min()) / (img.max() - img.min() + 1e-8)

        mask = np.load(mask_path).astype(np.float32)
        mask = (mask > 0).astype(np.float32)

        img = cv2.resize(img, (256, 256))
        mask = cv2.resize(mask, (256, 256), interpolation=cv2.INTER_NEAREST)

        img = torch.tensor(img).unsqueeze(0).float()
        mask = torch.tensor(mask).unsqueeze(0).float()

        return img, mask


dataset = CleanMaskDataset(IMG_DIR, MASK_DIR)
loader = DataLoader(dataset, batch_size=4, shuffle=True)

# 先檢查 GT
img, mask = dataset[0]

plt.figure(figsize=(8,4))
plt.subplot(1,2,1)
plt.title("Input")
plt.imshow(img[0], cmap="gray")
plt.axis("off")

plt.subplot(1,2,2)
plt.title("Clean GT")
plt.imshow(mask[0], cmap="gray")
plt.axis("off")
plt.show()


class UNet(nn.Module):
    def __init__(self):
        super().__init__()

        def block(in_c, out_c):
            return nn.Sequential(
                nn.Conv2d(in_c, out_c, 3, padding=1),
                nn.ReLU(inplace=True),
                nn.Conv2d(out_c, out_c, 3, padding=1),
                nn.ReLU(inplace=True)
            )

        self.enc1 = block(1, 16)
        self.enc2 = block(16, 32)
        self.enc3 = block(32, 64)

        self.pool = nn.MaxPool2d(2)

        self.up2 = nn.ConvTranspose2d(64, 32, 2, stride=2)
        self.dec2 = block(64, 32)

        self.up1 = nn.ConvTranspose2d(32, 16, 2, stride=2)
        self.dec1 = block(32, 16)

        self.out = nn.Conv2d(16, 1, 1)

    def forward(self, x):
        e1 = self.enc1(x)
        e2 = self.enc2(self.pool(e1))
        e3 = self.enc3(self.pool(e2))

        d2 = self.up2(e3)
        d2 = torch.cat([d2, e2], dim=1)
        d2 = self.dec2(d2)

        d1 = self.up1(d2)
        d1 = torch.cat([d1, e1], dim=1)
        d1 = self.dec1(d1)

        return self.out(d1)


device = "cuda" if torch.cuda.is_available() else "cpu"
print("device:", device)

model = UNet().to(device)
loss_fn = nn.BCEWithLogitsLoss()
opt = torch.optim.Adam(model.parameters(), lr=1e-3)

for epoch in range(30):
    model.train()
    total = 0

    for x, y in loader:
        x, y = x.to(device), y.to(device)

        pred = model(x)
        loss = loss_fn(pred, y)

        opt.zero_grad()
        loss.backward()
        opt.step()

        total += loss.item()

    print(f"epoch {epoch+1} loss {total:.4f}")


# 看 prediction
model.eval()

idx = min(20, len(dataset)-1)
img, mask = dataset[idx]
img_input = img.unsqueeze(0).to(device)

with torch.no_grad():
    pred = model(img_input)

pred = torch.sigmoid(pred)[0,0].cpu().numpy()

print("pred range:", pred.min(), pred.max())

plt.figure(figsize=(12,4))

plt.subplot(1,3,1)
plt.title("Input")
plt.imshow(img[0], cmap="gray")
plt.axis("off")

plt.subplot(1,3,2)
plt.title("Clean GT")
plt.imshow(mask[0], cmap="gray")
plt.axis("off")

plt.subplot(1,3,3)
plt.title("Prediction")
plt.imshow(pred > 0.3, cmap="gray")
plt.axis("off")

plt.show()
```
